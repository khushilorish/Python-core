```python
import pandas as pd
import numpy as np

# Create a sample DataFrame
data = {
    'CustomerID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112],
    'City': ['New York', 'Los Angeles', 'Chicago', 'New York', 'Los Angeles', 'Chicago', 'New York', 'Los Angeles', 'Chicago', 'New York', 'Los Angeles', 'Chicago'],
    'PurchaseAmount': [150.50, 220.75, 95.00, 310.20, 180.00, 450.50, 75.90, 290.00, 120.00, 550.00, 210.00, 300.00],
    'Category': ['Electronics', 'Clothing', 'Home Goods', 'Electronics', 'Clothing', 'Home Goods', 'Electronics', 'Clothing', 'Home Goods', 'Electronics', 'Clothing', 'Home Goods']
}
df_customers = pd.DataFrame(data)

print("Original DataFrame:")
print(df_customers)
```

```python
#  Filtering for Customers in 'New York'
ny_data = df_customers[df_customers['City'].str.contains('New York')]
print("Customers in New York:")
print(ny_data)
```

```python
# Filtering for Purchases Above $200
large_purchases = df_customers[df_customers['PurchaseAmount']>200]
print("Customer Data who Purchased above $200: ")
print(large_purchases)
```

```python
#  Combining Filters - Customers in 'Los Angeles' with Purchases > $200
la_high_value = df_customers[(df_customers['City'] == 'Los Angeles') & (df_customers['PurchaseAmount'] > 200)]

print("Customers in Los Angeles with purchases > $200:")
print(la_high_value)
```

```python
# Filter for customers in 'Chicago' OR 'Los Angeles'
chicago_or_la = df_customers[df_customers['City'].isin(['Chicago', 'Los Angeles'])]

print("Customers in Chicago or Los Angeles:")
print(chicago_or_la)
```
